```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet

models = {
    "Ridge" : Ridge(alpha=1.0),
    "Lasso" : Lasso(alpha = 1.0),
    "ElasticNet" : ElasticNet(alpha = 0.01),
}

print(models)
print(models["Ridge"])

for name, model in models.items():
  print(f"Model name: {name} -> Object: {model}")
```

Program 6: Conditional Statements
Write a Python program to classify a student's CGPA into High, Mid, or Low categories using if-elif-else.
Write a program to check whether missing values are present in a dataset using conditional statements.
Write a program to determine whether a student is eligible for placement based on CGPA.
Write a program that uses if-elif-else to assign grades based on marks.

```python
# --- IF / ELIF / ELSE ---

cgpa = 8.86

if cgpa >= 9:
  tier = "High"
elif cgpa >= 7:
  tier = "Mid"
else:
  tier = "Low"

print(f"CGPA : {cgpa} falls into tier : {tier}")

missing_count = 0
if missing_count == 0:
  print("No missing values")
else:
  print(f"Missing values: {missing_count}")
```

Program 7: Loops
Write a Python program to display all elements in a list using a for loop.
Write a program to calculate the average of numbers stored in a list.
Write a program to display only the values greater than a given threshold using a loop.
Write a program to find the maximum and minimum values in a list using loops.
Write a program to count the number of elements satisfying a given condition using a loop.

```python
numbers = [10, 20, 30, 40, 50, 5, 15, 25, 35, 45]

# Display all elements in a list using a for loop
print("1. Displaying all elements in the list:")
for num in numbers:
    print(num)

# Calculate the average of numbers stored in a list
print("\n2. Calculating the average of numbers:")
total = 0
for num in numbers:
    total += num
average = total / len(numbers)
print(f"The average of the numbers is: {average}")

# Display only the values greater than a given threshold using a loop
print("\n3. Displaying values greater than a threshold (30):")
threshold = 30
for num in numbers:
    if num > threshold:
        print(num)

# Find the maximum and minimum values in a list using loops
print("\n4. Finding the maximum and minimum values:")
if numbers:
    maximum = numbers[0]
    minimum = numbers[0]
    for num in numbers:
        if num > maximum:
            maximum = num
        if num < minimum:
            minimum = num
    print(f"Maximum value: {maximum}")
    print(f"Minimum value: {minimum}")
else:
    print("List is empty, cannot find max/min.")

# Count the number of elements satisfying a given condition using a loop
print("\n5. Counting elements greater than a threshold (20):")
count_greater_than_20 = 0
condition_threshold = 20
for num in numbers:
    if num > condition_threshold:
        count_greater_than_20 += 1
print(f"Number of elements greater than {condition_threshold}: {count_greater_than_20}")
```
